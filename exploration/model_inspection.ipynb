# Steerling model structure on Modal

This submits a detached L40S job that directly loads the pinned Hugging Face model and prints its PyTorch structure. Set `PROCEED=True` only after the experiment card is approved.

```python
import subprocess
import sys
from pathlib import Path

PROCEED = False
if not PROCEED:
    raise RuntimeError("Set PROCEED=True only after approving the experiment card.")

repo_root = next(path for path in [Path.cwd(), *Path.cwd().parents] if (path / "AGENTS.md").is_file())
entrypoint = repo_root / "exploration" / "modal_model_inspection.py"
command = [
    sys.executable, "-m", "modal", "run", "--detach", str(entrypoint), "--no-dry-run"
]
completed = subprocess.run(command, cwd=repo_root, check=True, text=True, capture_output=True)
print(completed.stdout)
```

Open the Modal URL printed above to view logs. The complete structure and summary are persisted on the `sterling-outputs` Volume under `exploration/runs/<run_id>`.
