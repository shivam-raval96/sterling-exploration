# Steerling model structure on Modal

Submit a detached L40S job that directly loads the pinned Hugging Face model and prints its PyTorch structure. Run cells individually from top to bottom.

## 1. Imports

```python
import subprocess
import sys
from pathlib import Path
```

## 2. Configuration

Keep the approval gate false until the experiment card has an explicit `proceed`.

```python
PROCEED = False
DETACHED = True
```

## 3. Resolve the repository and entrypoint

```python
search_paths = [Path.cwd(), *Path.cwd().parents]
repo_root = next(
    (path for path in search_paths if (path / "AGENTS.md").is_file()),
    None,
)
if repo_root is None:
    raise FileNotFoundError("Could not find the sterling-exploration repository root.")

entrypoint = repo_root / "exploration" / "modal_model_inspection.py"
if not entrypoint.is_file():
    raise FileNotFoundError(entrypoint)

print("Repository:", repo_root)
print("Entrypoint:", entrypoint)
```

## 4. Build and preview the command

This cell does not contact Modal.

```python
command = [sys.executable, "-m", "modal", "run"]
if DETACHED:
    command.append("--detach")
command.extend([str(entrypoint), "--no-dry-run"])
print(" ".join(command))
```

## 5. Approval gate

This cell must pass before submission.

```python
if not PROCEED:
    raise RuntimeError(
        "Set PROCEED=True only after approving exploration/EXPERIMENT_CARD.md."
    )
print("Approval gate passed.")
```

## 6. Submit to Modal

This is the only cell that launches remote compute.

```python
completed = subprocess.run(
    command,
    cwd=repo_root,
    check=True,
    text=True,
    capture_output=True,
)
```

## 7. Show submission details

```python
print(completed.stdout)
if completed.stderr:
    print(completed.stderr)
```

Open the Modal URL printed above to inspect logs. Persistent artifacts are written to `sterling-outputs/exploration/runs/<run_id>`.
